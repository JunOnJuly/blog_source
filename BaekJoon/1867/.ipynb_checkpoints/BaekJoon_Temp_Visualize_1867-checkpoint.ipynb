```python
# 이분 매칭
def bimatch(idx):
    # ----------------------------------------------------------------#
    plt.clf()
    plt.figure()
    fig, ax = plt.subplots(figsize=(10, 10))
    plt.xlim([-1, 11])
    plt.ylim([-1, n+1])
    plt.xticks([], [])
    plt.yticks([], [])
    plt.gca().spines['right'].set_visible(False)
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['left'].set_visible(False)
    plt.gca().spines['bottom'].set_visible(False)

    for i in range(n):
        for j in range(2):
            if j==0:
                ax.add_patch(
                   patches.Rectangle(
                      (j-0.4, i-0.4),
                      0.8, 0.8,
                      edgecolor = 'black',
                      facecolor = 'white',
                      fill=True,
                   ))
                ax.text(j-(len(str(i))/2)/10, i-0.1, i)
                
            else:
                ax.add_patch(
                   patches.Rectangle(
                      (j+9-0.4, i-0.4),
                      0.8, 0.8,
                      edgecolor = 'black',
                      facecolor = 'white',
                      fill=True,
                   ))
                ax.text(j+9-(len(str(i))/2)/10, i-0.1, i)
                
    ax.add_patch(
       patches.Rectangle(
          (-0.4, idx-0.4),
          0.8, 0.8,
          edgecolor = 'blue',
          facecolor = 'white',
          fill=True,
       ))
    
    ax.text(-(len(str(idx))/2)/10, idx-0.1, idx)
        
    for i, lst in enumerate(bi_graph):
        for j in lst:
            ax.plot([0.5, 9.5], [i, j], c='black')
            
    for j, i in enumerate(connected):
        if i >= 0:
            ax.plot([0.5, 9.5], [i, j], c='blue', linewidth=5)
        
    display.clear_output(wait=True)
    display.display(plt.gcf())
    fig_num = len(os.listdir(path))
    fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
    # ----------------------------------------------------------------#
    
    # 방문했으면 패스
    if visited[idx]:
        # ----------------------------------------------------------------#
        plt.clf()
        plt.figure()
        fig, ax = plt.subplots(figsize=(10, 10))
        plt.xlim([-1, 11])
        plt.ylim([-1, n+1])
        plt.xticks([], [])
        plt.yticks([], [])
        plt.gca().spines['right'].set_visible(False)
        plt.gca().spines['top'].set_visible(False)
        plt.gca().spines['left'].set_visible(False)
        plt.gca().spines['bottom'].set_visible(False)

        for i in range(n):
            for j in range(2):
                if j==0:
                    ax.add_patch(
                       patches.Rectangle(
                          (j-0.4, i-0.4),
                          0.8, 0.8,
                          edgecolor = 'black',
                          facecolor = 'white',
                          fill=True,
                       ))
                    ax.text(j-(len(str(i))/2)/10, i-0.1, i)

                else:
                    ax.add_patch(
                       patches.Rectangle(
                          (j+9-0.4, i-0.4),
                          0.8, 0.8,
                          edgecolor = 'black',
                          facecolor = 'white',
                          fill=True,
                       ))
                    ax.text(j+9-(len(str(i))/2)/10, i-0.1, i)
        
        ax.add_patch(
           patches.Rectangle(
              (-0.4, idx-0.4),
              0.8, 0.8,
              edgecolor = 'red',
              facecolor = 'white',
              fill=True,
           ))
        ax.text(-(len(str(idx))/2)/10, idx-0.1, idx)
                
        for i, lst in enumerate(bi_graph):
            for j in lst:
                ax.plot([0.5, 9.5], [i, j], c='black')
                
        for j, i in enumerate(connected):
            if i >= 0:
                ax.plot([0.5, 9.5], [i, j], c='blue', linewidth=5)
            
        display.clear_output(wait=True)
        display.display(plt.gcf())
        fig_num = len(os.listdir(path))
        fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
        # ----------------------------------------------------------------#
        
        return False
    # 방문 체크
    visited[idx] = True

    # 연결될 수 있는 노드 체크
    for node in bi_graph[idx]:
        # ----------------------------------------------------------------#
        plt.clf()
        plt.figure()
        fig, ax = plt.subplots(figsize=(10, 10))
        plt.xlim([-1, 11])
        plt.ylim([-1, n+1])
        plt.xticks([], [])
        plt.yticks([], [])
        plt.gca().spines['right'].set_visible(False)
        plt.gca().spines['top'].set_visible(False)
        plt.gca().spines['left'].set_visible(False)
        plt.gca().spines['bottom'].set_visible(False)

        for i in range(n):
            for j in range(2):
                if j==0:
                    ax.add_patch(
                       patches.Rectangle(
                          (j-0.4, i-0.4),
                          0.8, 0.8,
                          edgecolor = 'black',
                          facecolor = 'white',
                          fill=True,
                       ))
                    ax.text(j-(len(str(i))/2)/10, i-0.1, i)

                else:
                    ax.add_patch(
                       patches.Rectangle(
                          (j+9-0.4, i-0.4),
                          0.8, 0.8,
                          edgecolor = 'black',
                          facecolor = 'white',
                          fill=True,
                       ))
                    ax.text(j+9-(len(str(i))/2)/10, i-0.1, i)

        ax.add_patch(
           patches.Rectangle(
              (-0.4, idx-0.4),
              0.8, 0.8,
              edgecolor = 'blue',
              facecolor = 'white',
              fill=True,
           ))
        ax.text(-(len(str(idx))/2)/10, idx-0.1, idx)

        ax.add_patch(
           patches.Rectangle(
              (10-0.4, node-0.4),
              0.8, 0.8,
              edgecolor = 'blue',
              facecolor = 'white',
              fill=True,
           ))
        ax.text(10-(len(str(idx))/2)/10, idx-0.1, idx)

        for i, lst in enumerate(bi_graph):
            for j in lst:
                ax.plot([0.5, 9.5], [i, j], c='black')

        for j, i in enumerate(connected):
            if i >= 0:
                ax.plot([0.5, 9.5], [i, j], c='blue', linewidth=5)

        ax.plot([0.5, 9.5], [idx, node], c='black', linewidth=5)
            
        display.clear_output(wait=True)
        display.display(plt.gcf())
        fig_num = len(os.listdir(path))
        fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
        # ----------------------------------------------------------------#
        # 연결되지 않았거나
        # 연결된 노드와 연결된 노드가 다른 노드를 선택할 수 있으면
        if connected[node] < 0 or bimatch(connected[node]):
            # 연결
            connected[node] = idx
            # ----------------------------------------------------------------#
            plt.clf()
            plt.figure()
            fig, ax = plt.subplots(figsize=(10, 10))
            plt.xlim([-1, 11])
            plt.ylim([-1, n+1])
            plt.xticks([], [])
            plt.yticks([], [])
            plt.gca().spines['right'].set_visible(False)
            plt.gca().spines['top'].set_visible(False)
            plt.gca().spines['left'].set_visible(False)
            plt.gca().spines['bottom'].set_visible(False)

            for i in range(n):
                for j in range(2):
                    if j==0:
                        ax.add_patch(
                           patches.Rectangle(
                              (j-0.4, i-0.4),
                              0.8, 0.8,
                              edgecolor = 'black',
                              facecolor = 'white',
                              fill=True,
                           ))
                        ax.text(j-(len(str(i))/2)/10, i-0.1, i)

                    else:
                        ax.add_patch(
                           patches.Rectangle(
                              (j+9-0.4, i-0.4),
                              0.8, 0.8,
                              edgecolor = 'black',
                              facecolor = 'white',
                              fill=True,
                           ))
                        ax.text(j+9-(len(str(i))/2)/10, i-0.1, i)
            
            ax.add_patch(
               patches.Rectangle(
                  (-0.4, idx-0.4),
                  0.8, 0.8,
                  edgecolor = 'blue',
                  facecolor = 'white',
                  fill=True,
               ))
            ax.text(-(len(str(idx))/2)/10, idx-0.1, idx)
            
            ax.add_patch(
               patches.Rectangle(
                  (10-0.4, node-0.4),
                  0.8, 0.8,
                  edgecolor = 'blue',
                  facecolor = 'white',
                  fill=True,
               ))
            ax.text(10-(len(str(idx))/2)/10, idx-0.1, idx)
            
            for i, lst in enumerate(bi_graph):
                for j in lst:
                    ax.plot([0.5, 9.5], [i, j], c='black')
                    
            for j, i in enumerate(connected):
                if i >= 0:
                    ax.plot([0.5, 9.5], [i, j], c='blue', linewidth=5)
                
            display.clear_output(wait=True)
            display.display(plt.gcf())
            fig_num = len(os.listdir(path))
            fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
            # ----------------------------------------------------------------#
            return True
        else:
            # ----------------------------------------------------------------#
            plt.clf()
            plt.figure()
            fig, ax = plt.subplots(figsize=(10, 10))
            plt.xlim([-1, 11])
            plt.ylim([-1, n+1])
            plt.xticks([], [])
            plt.yticks([], [])
            plt.gca().spines['right'].set_visible(False)
            plt.gca().spines['top'].set_visible(False)
            plt.gca().spines['left'].set_visible(False)
            plt.gca().spines['bottom'].set_visible(False)

            for i in range(n):
                for j in range(2):
                    if j==0:
                        ax.add_patch(
                           patches.Rectangle(
                              (j-0.4, i-0.4),
                              0.8, 0.8,
                              edgecolor = 'black',
                              facecolor = 'white',
                              fill=True,
                           ))
                        ax.text(j-(len(str(i))/2)/10, i-0.1, i)

                    else:
                        ax.add_patch(
                           patches.Rectangle(
                              (j+9-0.4, i-0.4),
                              0.8, 0.8,
                              edgecolor = 'black',
                              facecolor = 'white',
                              fill=True,
                           ))
                        ax.text(j+9-(len(str(i))/2)/10, i-0.1, i)
            
            ax.add_patch(
               patches.Rectangle(
                  (-0.4, idx-0.4),
                  0.8, 0.8,
                  edgecolor = 'blue',
                  facecolor = 'white',
                  fill=True,
               ))
            ax.text(-(len(str(idx))/2)/10, idx-0.1, idx)
            
            ax.add_patch(
               patches.Rectangle(
                  (10-0.4, node-0.4),
                  0.8, 0.8,
                  edgecolor = 'blue',
                  facecolor = 'white',
                  fill=True,
               ))
            ax.text(10-(len(str(idx))/2)/10, idx-0.1, idx)
            
            for i, lst in enumerate(bi_graph):
                for j in lst:
                    ax.plot([0.5, 9.5], [i, j], c='black')
                    
            for j, i in enumerate(connected):
                if i >= 0:
                    ax.plot([0.5, 9.5], [i, j], c='blue', linewidth=5)
            
            ax.plot([0.5, 9.5], [connected[node], node], c='red', linewidth=5)
            
            display.clear_output(wait=True)
            display.display(plt.gcf())
            fig_num = len(os.listdir(path))
            fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
            # ----------------------------------------------------------------#
    return False


def solution(n):
    
    # ----------------------------------------------------------------#
    if not os.path.exists(path):
        os.mkdir(path)
        
    plt.clf()
    plt.figure()
    fig, ax = plt.subplots(figsize=(10, 10))
    plt.xlim([-1, 11])
    plt.ylim([-1, n+1])
    plt.xticks([], [])
    plt.yticks([], [])
    plt.gca().spines['right'].set_visible(False)
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['left'].set_visible(False)
    plt.gca().spines['bottom'].set_visible(False)

    for i in range(n):
        for j in range(2):
            if j==0:
                ax.add_patch(
                   patches.Rectangle(
                      (j-0.4, i-0.4),
                      0.8, 0.8,
                      edgecolor = 'black',
                      facecolor = 'white',
                      fill=True,
                   ))
                ax.text(j-(len(str(i))/2)/10, i-0.1, i)
                
            else:
                ax.add_patch(
                   patches.Rectangle(
                      (j+9-0.4, i-0.4),
                      0.8, 0.8,
                      edgecolor = 'black',
                      facecolor = 'white',
                      fill=True,
                   ))
                ax.text(j+9-(len(str(i))/2)/10, i-0.1, i)
    
    for i, lst in enumerate(bi_graph):
        for j in lst:
            ax.plot([0.5, 9.5], [i, j], c='black')

    display.clear_output(wait=True)
    display.display(plt.gcf())
    fig_num = len(os.listdir(path))
    fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
    # ----------------------------------------------------------------#
    
    # 결과
    result = 0
    # 연결 기록(열노드)
    global connected
    connected = [-1 for _ in range(n)]
    # 행노드 순회하며 이분매칭
    for idx in range(n):
        # 방문 기록(행노드)
        global visited
        visited = [False for _ in range(n)]
        if bimatch(idx):
            result += 1

    return result
```

```python
n = random.randint(6, 10)
k = random.randint((n**2)//3, (n**2)//2)

random_comb = list(product(range(n), repeat=2))
random_nums = random.choices(range(n**2), k=k)

bi_graph = [[] for _ in range(n)]
for num in random_nums:
    r, c = random_comb[num]
    bi_graph[r-1].append(c)

print(solution(n))
```

```python
plt.figure(figsize=(10, 10))
for r, cs in enumerate(bi_graph):
    for c in cs:
        plt.scatter(c, r, s=500, c='k')
for i in range(1, n):
    plt.plot([i-0.5, i-0.5], [0-0.5, n-0.5], 'k')
    plt.plot([0-0.5, n-0.5], [i-0.5, i-0.5], 'k')
plt.xticks([],[])
plt.yticks([],[])
plt.show()
```
