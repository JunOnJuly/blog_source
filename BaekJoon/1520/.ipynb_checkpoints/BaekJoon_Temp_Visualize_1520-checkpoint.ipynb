```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import heapq
from time import sleep
from IPython import display
import sys
import os
from PIL import Image

path = '../IMAGES/GIFS/1520'
```

```python
def in_hq(hq, idx):
    for _, [x, y] in hq:
        if x == idx[0] and y == idx[1]:
            return True
    else:
        return False


def solution(M, N, data_map):
    # ----------------------------------------------------------------#
    fig, ax = plt.subplots()
    if not os.path.exists(path):
        os.mkdir(path)
    # ----------------------------------------------------------------#
    
    # 이동 가이드
    move_guide = [[0, 1], [0, -1], [1, 0], [-1, 0]]
    # 방문 횟수 리스트
    visited = [[0 for _ in range(N)] for __ in range(M)]
    visited[0][0] = 1
    # 큐
    hq = [[-data_map[0][0], [0, 0]]]
    # 순회
    while hq:
        # 현재 높이, 현재 인덱스
        now_height, [now_x, now_y] = heapq.heappop(hq)
        now_height *= -1
        
        # ----------------------------------------------------------------#
        plt.clf()
        fig, ax = plt.subplots()
        plt.xlim([-1, N])
        plt.ylim([-1, M])
        plt.xticks([], [])
        plt.yticks([], [])
        plt.gca().spines['right'].set_visible(False)
        plt.gca().spines['top'].set_visible(False)
        plt.gca().spines['left'].set_visible(False)
        plt.gca().spines['bottom'].set_visible(False)
            
        for i in range(M):
            for j in range(N):
                if in_hq(hq, [i, j]):
                    ax.add_patch(
                       patches.Rectangle(
                          (j-0.4, M-i-1-0.4),
                          0.8, 0.8,
                          edgecolor = 'blue',
                          facecolor = 'white',
                          fill=True,
                       ))
                elif now_x == i and now_y == j:
                    ax.add_patch(
                       patches.Rectangle(
                          (j-0.4, M-i-1-0.4),
                          0.8, 0.8,
                          edgecolor = 'red',
                          facecolor = 'white',
                          fill=True,
                       ))
                else:
                    ax.add_patch(
                       patches.Rectangle(
                          (j-0.4, M-i-1-0.4),
                          0.8, 0.8,
                          edgecolor = 'black',
                          facecolor = 'white',
                          fill=True,
                       ))
                height = data_map[M-i-1][j]
                cnt = visited[M-i-1][j]
                ax.text(j-(len(str(height))/2)/10, i, height)
                ax.text(j-(len(str(cnt))/2)/10, i-0.2, cnt)
                
        display.clear_output(wait=True)
        display.display(plt.gcf())
        fig_num = len(os.listdir(path))
        fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
        # ----------------------------------------------------------------#
        
        # 이동 가능한 위치 순회
        for x, y in move_guide:
            # 다음 인덱스
            next_x, next_y = now_x + x, now_y + y
            # 이동 가능한 위치면
            if (next_x >= 0 and next_x < M) and (next_y >= 0 and next_y < N):
                # 다음 위치의 높이가 더 낮으면
                if data_map[next_x][next_y] < now_height:
                    # 방문한 적이 없다면
                    if not visited[next_x][next_y]:
                        # 큐에 삽입
                        heapq.heappush(hq, [-data_map[next_x][next_y], [next_x, next_y]])
                    # 방문 횟수 최신화
                    visited[next_x][next_y] += visited[now_x][now_y]
                    
                    # ----------------------------------------------------------------#
                    plt.clf()
                    fig, ax = plt.subplots()
                    plt.xlim([-1, N])
                    plt.ylim([-1, M])
                    plt.xticks([], [])
                    plt.yticks([], [])
                    plt.gca().spines['right'].set_visible(False)
                    plt.gca().spines['top'].set_visible(False)
                    plt.gca().spines['left'].set_visible(False)
                    plt.gca().spines['bottom'].set_visible(False)

                    for i in range(M):
                        for j in range(N):
                            if in_hq(hq, [i, j]):
                                ax.add_patch(
                                   patches.Rectangle(
                                      (j-0.4, M-i-1-0.4),
                                      0.8, 0.8,
                                      edgecolor = 'blue',
                                      facecolor = 'white',
                                      fill=True,
                                   ))
                            elif now_x == i and now_y == j:
                                ax.add_patch(
                                   patches.Rectangle(
                                      (j-0.4, M-i-1-0.4),
                                      0.8, 0.8,
                                      edgecolor = 'red',
                                      facecolor = 'white',
                                      fill=True,
                                   ))
                            else:
                                ax.add_patch(
                                   patches.Rectangle(
                                      (j-0.4, M-i-1-0.4),
                                      0.8, 0.8,
                                      edgecolor = 'black',
                                      facecolor = 'white',
                                      fill=True,
                                   ))
                            height = data_map[M-i-1][j]
                            cnt = visited[M-i-1][j]
                            ax.text(j-(len(str(height))/2)/10, i, height)
                            ax.text(j-(len(str(cnt))/2)/10, i-0.2, cnt)

                    display.clear_output(wait=True)
                    display.display(plt.gcf())
                    fig_num = len(os.listdir(path))
                    fig.savefig(f'{path}/{fig_num}.png', dpi=fig.dpi)
                    # ----------------------------------------------------------------#


    return visited[M-1][N-1]
```

```python
# # 입력
# M, N = map(int, input().strip().split())
# data_map = [list(map(int, input().strip().split())) for _ in range(M)]

# print(solution(M, N, data_map))
# 입력
M, N = 4, 5
data_map = [
[50, 45, 37, 32, 30],
[35, 50, 40, 20, 25],
[30, 30, 25, 17, 28],
[27, 24, 22, 15, 10]]
solution(M, N, data_map)
```
